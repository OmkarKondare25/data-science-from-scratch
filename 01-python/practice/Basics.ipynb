```python
## Comparison Operators

a=10;b=10

print(a==b) ## equal to
print(a!=b) ## not equal to
print(a>b) ## greater than
print(a<b) ## less than
print(a<=b) ## greater than or equal to
print(a>=b) ## less than or equal to
```

```python
## Logical Operators

a = True
b = False

print(a and b) ## if both are equal the returns true else false
print(a or b) ## if one of them is true returns true else false
print(not b) ## inverse of variable
```
